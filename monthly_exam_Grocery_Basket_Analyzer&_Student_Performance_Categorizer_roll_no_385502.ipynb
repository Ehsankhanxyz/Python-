```python
#PROJECT 01: GROCERY BASKET ANALYIZER
#1 TO CHECK FOR STOCKS LIMIT
basket_items_1 = [('item1',2), ('item2', 3), ('item3', 4),('item5',5)]
#to check for stock limits
if basket_items_1[0][1]>10:
    print(f' {basket_items_1[0][1]} exceeds the purchase limit!')
else:
    print (f'Your haved purchased {basket_items_1[0][0]} and quantity: {basket_items_1[0][1]}')
```

```python
#2 TO CALCULATE TOTAL QUANTITY OF ITEMS PER BASKET
total_basket1_items = 0
for items in basket_items_1:
    total_basket1_items += items[1]
print(f' Total quanitiy of items per basket is {total_basket1_items}')
```

```python
#3 TO STORE ALL UNIQUE ITEMS NAMES FROM ALL BASKETS IN A SET

basket_items_1 = [('item1',2), ('item2', 3), ('item3', 4),('item5',5)]
basket_items_2 = [('item11',3), ('item12', 2), ('item13', 8),('item15',7)]
basket_items_3 = [('item21',6), ('item22', 8), ('item23', 9),('item25',3)]
all_baskets_unique_items = []

for i in basket_items_1:
    all_baskets_unique_items.append(i[0])
for i in basket_items_2:
    all_baskets_unique_items.append(i[0])
for i in basket_items_3:
    all_baskets_unique_items.append(i[0])
print(set(all_baskets_unique_items))
```

```python
# 1. Classify each student into one of the following performance categories using conditional statements:
# ○ Excellent (score ≥ 85)
# ○ Good (70 ≤ score < 85)

# ○ Average (50 ≤ score < 70)
# ○ Poor (score < 50)

class_students_list = [('student1', 55), ('student2', 90), ('student3', 65), ('student4', 45), ('student5', 70)]
excellent_students= []
good_students = []
avg_students = []
poor_students = []

#using conditional statments
#for student1
if class_students_list[0][1]>= 85:
  excellent_students.append(class_students_list[0])
elif (class_students_list[0][1]<85 and class_students_list[0][1]>= 70):
  good_students.append(class_students_list[0])
elif (class_students_list[0][1]<70 and class_students_list[0][1]>=50):
  avg_students.append(class_students_list[0])
else:
    poor_students.append(class_students_list[0])
#for student2
if class_students_list[1][1]>= 85:
  excellent_students.append(class_students_list[1])
elif (class_students_list[1][1]<85 and class_students_list[1][1]>= 70):
  good_students.append(class_students_list[1])
elif (class_students_list[1][1]<70 and class_students_list[1][1]>=50):
  avg_students.append(class_students_list[1])
else:
    poor_students.append(class_students_list[1])


print (f' Excellent Students are: {excellent_students} \n Good Students are: {good_students} \n Average Students are: {avg_students}\n Poor Students are: {poor_students}')
```

```python
#2. Display the students in each category using a for loop.
#for loop to classify student into criteria list
class_students_list = [('student1', 55), ('student2', 90), ('student3', 65), ('student4', 45), ('student5', 70)]
excellent_students= []
good_students = []
avg_students = []
poor_students = []

for i in class_students_list:
  if i[1]>= 85:
    excellent_students.append(i[0])
  elif (i[1]<85 and i[1]>= 70):
    good_students.append(i[0])
  elif (i[1]<70 and i[1]>=50):
    avg_students.append(i[0])
  else:
    poor_students.append(i[0])
print (f' Excellent Students are: {excellent_students} \n Good Students are: {good_students} \n Average Students are: {avg_students}\n Poor Students are: {poor_students}')
```
